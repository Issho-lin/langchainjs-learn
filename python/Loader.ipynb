```python
from langchain_community.document_loaders import DirectoryLoader, TextLoader, PyPDFLoader
from pprint import pprint

def get_loader(file_path: str):
  if (file_path.endswith('.txt')):
    return TextLoader(file_path=file_path)
  else:
    return PyPDFLoader(file_path=file_path, mode='single')


loader = DirectoryLoader(
  path='../documents/',
  glob=['*.txt', '*.pdf'],
  loader_cls=get_loader # type: ignore
)

rst = loader.load()

pprint(rst)
```

```python
from langchain_community.document_loaders import Docx2txtLoader

loader = Docx2txtLoader(file_path='../documents/data.docx')

rst = loader.load()

print(rst)
```

```python
from langchain_community.document_loaders import WebBaseLoader, AsyncHtmlLoader
from bs4.filter import SoupStrainer

loader = WebBaseLoader(
    web_path='https://python.langchain.ac.cn/docs/introduction',
    bs_kwargs=dict(
        parse_only=SoupStrainer(name='p')
    ),
)

rst = loader.load()

print(rst)
```
