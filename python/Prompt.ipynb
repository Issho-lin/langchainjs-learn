```python
# 动态参数
from langchain_core.prompts import PromptTemplate
import datetime

prompt = PromptTemplate.from_template(
    '现在是北京时间{time}，请告诉我今天{city}的天气。'
)

def get_time(show_time: bool = True) -> str:
    if show_time:
        return datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S')
    return datetime.datetime.now().strftime('%Y-%m-%d')

time_prompt = prompt.partial(time=lambda: get_time(show_time=True))
```

```python
time_prompt.format(city='深圳')
```

```python
# 组合prompt
from langchain_core.prompts import PromptTemplate, PipelinePromptTemplate
import datetime

prompt = PromptTemplate.from_template(
    '''
    你是一个{job}，现在是{time}，
    请根据你掌握的最新的知识，帮我完成以下需求：
    {task}
    '''
)

date_prompt = PromptTemplate.from_template(
    '{city}时间{time}'
)

def get_time(show_time: bool = True) -> str:
    if show_time:
        return datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S')
    return datetime.datetime.now().strftime('%Y-%m-%d')

date_partial = date_prompt.partial(time=lambda: get_time(show_time=True))

task_prompt = PromptTemplate.from_template(
    '''
    帮我生成一份{num}条{topic}的文章，
    要求文章的字数在{min_length}到{max_length}字之间，
    文章的风格要{style}。
    '''
)

compose_prompt = PipelinePromptTemplate(
    pipeline_prompts=[
        ('time', date_partial),
        ('task', task_prompt),
    ],
    final_prompt=prompt,
)

rst = compose_prompt.format(
    job='文案撰写员',
    num=3,
    topic='人工智能',
    min_length=1000,
    max_length=2000,
    style='严谨',
    city='北京',
)

print(rst)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "你是一个专业的翻译官，请把用户输入的信息从{input_language}翻译成{output_language}。"),
        ("human", "{text}"),
    ]
)

rst = prompt.format(input_language="中文", output_language="英文", text="Python是世界上最好的语言。")

print(rst)
```

```python
from langchain_core.prompts import ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate, AIMessagePromptTemplate, ChatMessagePromptTemplate

prompt = ChatPromptTemplate.from_messages([
    SystemMessagePromptTemplate.from_template("你是一个专业的翻译官，请把用户输入的信息从{input_language}翻译成{output_language}。"),
    HumanMessagePromptTemplate.from_template("{text}"),
])

rst = prompt.format(input_language="中文", output_language="英文", text="Python是世界上最好的语言。")

print(rst)
```
