```python
from langchain_openai import AzureChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

import os
import json

chat_model = AzureChatOpenAI(

    api_key=os.environ["AZURE_OPENAI_API_KEY"],

    azure_endpoint=os.environ["AZURE_OPENAI_ENDPOINT"],

    azure_deployment=os.environ["AZURE_OPENAI_DEPLOYMENT_NAME"],

    api_version=os.environ["AZURE_OPENAI_API_VERSION"],
)


# rst = chat_model.predict("你好")
rst = chat_model.invoke([
    SystemMessage(content="你是一名暴躁的Java程序员。"),
    HumanMessage(content="Python是世界上最好的语言。"),
    AIMessage(content="Java才是世界上最好的语言。"),
    HumanMessage(content="Python是世界上最好的语言。"),
])

# rst = chat_model.invoke([
#     ('system', '你是一名暴躁的Java程序员。'),
#     ('human', 'Python是世界上最好的语言。'),
#     ('ai', 'Java才是世界上最好的语言。'),
#     ('human', 'Python是世界上最好的语言。'),
# ])

# print(json.dumps(rst.model_dump(), indent=2, ensure_ascii=False))

rst = chat_model.batch([
    [HumanMessage(content="hello")],
    [HumanMessage(content="你好")],
])

print([item.content for item in rst])
```

```python
from langchain_openai import ChatOpenAI
import os

chat_model = ChatOpenAI(
    base_url=f'https://{os.environ["AZURE_OPENAI_API_INSTANCE_NAME"]}.openai.azure.com/openai/deployments/{os.environ["AZURE_OPENAI_DEPLOYMENT_NAME"]}',
    api_key=os.environ["AZURE_OPENAI_API_KEY"],
    default_query={
       "api-version": os.environ["AZURE_OPENAI_API_VERSION"] ,
    },
)

rst = chat_model.invoke(input='你好啊')
print(rst.content)
```
